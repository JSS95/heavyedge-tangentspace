# Vectorized profiles

Visualize vectorization result.

```
X = "../result/dimless/mean_profiles.csv"
y = "../_temp/umap/mean_profiles.csv"
color = "gap_to_thickness_ratio"
```

```
import pandas as pd

parameters = pd.read_csv(X).set_index("name")
features = pd.read_csv(y).set_index("name")
c = parameters[color].reindex(features.index)
```

```
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 5))
scatter = ax.scatter(
    features.values[:, 0], features.values[:, 1], c=c, cmap="viridis", s=35
)
fig.colorbar(scatter, ax=ax, label="gap_to_thickness_ratio")
ax.set(xlabel="UMAP 1", ylabel="UMAP 2", title="Vectorized profiles")
fig.tight_layout()
plt.show()
```
