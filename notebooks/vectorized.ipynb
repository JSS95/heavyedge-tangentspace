# Vectorized profiles

Visualize vectorization result.

```
X = "../result/dimless/mean_profiles.csv"
y = "../result/vectorized/mean_profiles.csv"
```

```
import matplotlib.pyplot as plt
import pandas as pd
import umap

features = pd.read_csv(y).set_index("name")
parameters = pd.read_csv(X).set_index("name")
gap_ratio = parameters["gap_to_thickness_ratio"].reindex(features.index)
if gap_ratio.isna().any():
    raise ValueError("Some feature vectors have no gap_to_thickness_ratio.")

embedding = umap.UMAP(n_components=2, random_state=42).fit_transform(
    features.to_numpy()
)

fig, ax = plt.subplots(figsize=(7, 5))
scatter = ax.scatter(
    embedding[:, 0], embedding[:, 1], c=gap_ratio, cmap="viridis", s=35
)
fig.colorbar(scatter, ax=ax, label="gap_to_thickness_ratio")
ax.set(xlabel="UMAP 1", ylabel="UMAP 2", title="Vectorized profiles")
fig.tight_layout()
plt.show()
```
